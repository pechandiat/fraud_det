```python
# Modificar features numericas que vienen como strings con comas
def to_float(x):
    return float(x.replace(",", ""))


cols = ["Q", "R", "Monto"]
df[cols] = df[cols].map(to_float)

```

```python
cols_to_drop = ['K']

categorical_cols = ['J']
numerical_cols = df.select_dtypes(include=["number"]).columns

data_cleaned = df.drop(columns=cols_to_drop)
data_cleaned['C_is_null'] = data_cleaned['C'].isna().astype(int)
data_cleaned.head()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OrdinalEncoder, StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, average_precision_score

TARGET = 'Fraude'

x = data_cleaned.drop(columns=[TARGET])
y = data_cleaned[TARGET]

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42, stratify=y)
```

```python
categorical = ["J"]
numerical_needs_imputation = ['C']
numerical = [col for col in x.columns if col not in categorical + numerical_needs_imputation]
RANDOM = 42
CACHE = ROOT / "cache"

transform = ColumnTransformer(
    [
        ("numerical", "passthrough", numerical),
        ("numerical_imputed", SimpleImputer(strategy='median'), numerical_needs_imputation),
        ("categorical", OneHotEncoder(sparse_output=False, handle_unknown='ignore'), categorical)
    ]
)

pipe = Pipeline(
    steps=[
        ('transform', transform),
        ('model', RandomForestClassifier(random_state=RANDOM, n_estimators=100, class_weight='balanced'))
    ])

pipe.fit(X_train, y_train)
```

```python
y_pred = pipe.predict(X_test)
y_proba = pipe.predict_proba(X_test)[:, 1]

print("=== Classification Report ===")
print(classification_report(y_test, y_pred, digits=3, zero_division=0))

print("=== Confusion Matrix ===")
cm = confusion_matrix(y_test, y_pred)
print(cm)

print(f"\nROC-AUC: {roc_auc_score(y_test, y_proba, ):.3f}")
print(f"PR-AUC (Average Precision): {average_precision_score(y_test, y_proba):.3f}")
```

```python
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
pr_auc = average_precision_score(y_test, y_proba)

print(
    "Random Forest - Precision: {:.4f}, Recall: {:.4f}, PR-AUC: {:.4f}".format(
        precision, recall, pr_auc
    )
)
```

```python
plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No Fraude", "Fraude"],
    yticklabels=["No Fraude", "Fraude"],
)
plt.xlabel("Predicción")
plt.ylabel("Real")
plt.title("Matriz de Confusión")
plt.show()
```

```python
from sklearn.model_selection import cross_validate, StratifiedKFold

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

scoring = {
    'f1':'f1',
    'recall':'recall',
    'precision':'precision',
    'roc_auc':'roc_auc',
    'pr_auc':'average_precision'
}

cv_results = cross_validate(pipe, x, y, cv=cv, scoring=scoring, return_train_score=False)

for metric in scoring:
    scores = cv_results[f'test_{metric}']
    print(f"{metric:12s}: {scores.mean():.3f} ± {scores.std():.3f} (folds: {[round(s,3) for s in scores]})")
```
